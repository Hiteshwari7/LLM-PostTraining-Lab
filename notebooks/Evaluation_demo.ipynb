```python
import torch
from huggingface_hub import login
from unsloth import FastLanguageModel

login()
```

```python
MODEL_NAME = "unsloth/Llama-3.2-3B-Instruct-bnb-4bit"

SFT_ADAPTER_REPO = "Hiteshwari7/posttraining-tutor-sft-adapter"
DPO_ADAPTER_REPO = "Hiteshwari7/postraining-tutor-dpo-adapter"

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=MODEL_NAME,
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
)

model.load_adapter(
    SFT_ADAPTER_REPO,
    adapter_name="sft"
)

model.load_adapter(
    DPO_ADAPTER_REPO,
    adapter_name="dpo"
)

FastLanguageModel.for_inference(model)

print("Adapters:", list(model.peft_config.keys()))
```

```python
import gradio as gr
import traceback

def gradio_generate(question, model_choice):
    try:
        if not question.strip():
            return "Please enter a question."

        # Select adapter
        if model_choice == "Base Model":
            model.disable_adapters()
        elif model_choice == "SFT Model":
            model.enable_adapters()
            model.set_adapter("sft")
        elif model_choice == "DPO Model":
            model.enable_adapters()
            model.set_adapter("dpo")

        # Tokenize
        inputs = tokenizer(
            question,
            return_tensors="pt"
        )

        # Move inputs to the same device as the model
        inputs = {k: v.to(model.device) for k, v in inputs.items()}

        # Generate
        with torch.inference_mode():
            outputs = model.generate(
                **inputs,
                max_new_tokens=200,
                do_sample=False,
                pad_token_id=tokenizer.eos_token_id,
            )

        # Only return generated answer, not the prompt
        input_length = inputs["input_ids"].shape[1]

        answer = tokenizer.decode(
            outputs[0][input_length:],
            skip_special_tokens=True
        )

        return answer.strip()

    except Exception as e:
        print("GRADIO ERROR:")
        traceback.print_exc()
        return f"ERROR: {type(e).__name__}: {str(e)}"


demo = gr.Interface(
    fn=gradio_generate,

    inputs=[
        gr.Textbox(
            label="Ask PostTraining Tutor",
            placeholder="e.g. What is LoRA and why is it useful?",
            lines=3
        ),

        gr.Radio(
            choices=[
                "Base Model",
                "SFT Model",
                "DPO Model"
            ],
            value="DPO Model",
            label="Select Model"
        )
    ],

    outputs=gr.Textbox(
        label="Answer",
        lines=12
    ),

    title="🤖 PostTraining Tutor",

    description=(
        "Compare Base, SFT, and DPO versions "
        "of Llama-3.2-3B-Instruct."
    ),

    examples=[
        ["What is LoRA and why is it useful?", "Base Model"],
        ["What is LoRA and why is it useful?", "SFT Model"],
        ["What is LoRA and why is it useful?", "DPO Model"],
    ],
)

demo.launch(share=True, debug=True)
```
